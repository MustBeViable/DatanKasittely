## 1. Business Understanding

We need to find correlation between product sales to increase revenue based on the data.

## 2. Data Understanding

The data has 100000 entries and 20 features, prod1-prod20, no null values and every feature is int64

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_csv('../datasets/drone_prod_groups.csv')
df
```

```python
print("Dataset shape:", df.shape)
```

The data has ID and 20 product columns which are int64s. The dataset doesnt have missing values

```python
print("Data types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())

print("\nNumber of duplicate rows:", df.duplicated().sum())
```

## 3. Data Preparation
Lets replace empty and 0 values with False and 1 values with True

```python
df = df.drop(columns='ID')
df = df.fillna(False)
df = df.apply(lambda x: x.map(lambda y: True if y == 1.0 else False))
df
```

## 4. Modeling

lets set min support to a value where meaningful relationships can be found

```python
frequent_itemsets = apriori(df, min_support=0.01, use_colnames=True)
frequent_itemsets
```

Lets generate association rules

```python
rules = association_rules(frequent_itemsets,metric="confidence",min_threshold=0.5)
rules = rules.sort_values(by=["confidence", "lift", "support"],ascending=[False, False, False])
rules
```

## 5. Evaluation

```python
rules_report = rules[["antecedents","consequents","support","confidence","lift"]].copy()
rules_report
```

Lets make the frozen sets more readable

```python
rules_report["antecedents"] = rules_report["antecedents"].apply(lambda x: ", ".join(sorted(x)))
rules_report["consequents"] = rules_report["consequents"].apply(lambda x: ", ".join(sorted(x)))
rules_report
```

Prod9 has good support, confidence and lift with groups {Prod15, Prod2}, {Prod15, Prod20}, {Prod15, Prod19, Prod20},{Prod12, Prod15},{Prod15, Prod7}.

### Deployment / Conclusion

Association rule analysis show prod9 having high cross selling with prod15, prod2, prod20, prod19, prod7 and prod12. Identified with groupings of {Prod15, Prod2}, {Prod15, Prod20}, {Prod15, Prod19, Prod20},{Prod12, Prod15},{Prod15, Prod7}.
