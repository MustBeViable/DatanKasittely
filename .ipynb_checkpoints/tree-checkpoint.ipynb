```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
phishing_websites = fetch_ucirepo(id=327) 
  
# data (as pandas dataframes) 
X = phishing_websites.data.features 
y = phishing_websites.data.targets 
#change y from from (x,1) to (x,)
y = y.squeeze()
# metadata 
#print(phishing_websites.metadata) 
  
# variable information 
print(phishing_websites.variables) 
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state = 1)

model = DecisionTreeClassifier(max_depth= 3 ,random_state=1)
model.fit(X_train, y_train)

fig = plt.figure(figsize = (15,15))
plot_tree(model, feature_names = X.keys(), class_names=['philsing','legitimate'])#dataset had [-1,1] -1 for philsing and 1 for legitimate no 0 for suspicious as stated in the provided article
plt.show()

preds = model.predict(X_test)
print(confusion_matrix(y_test, preds))
print()
print(classification_report(y_test, preds))
```

```python
#random forest
from sklearn.ensemble import RandomForestClassifier

results = []
for n_trees in [50,100,200]:
    for max_depth in [5,10,20, None]:
        for min_leaf in [1, 5, 20]:
            for max_samples in [0.5, 1.0]:
                for max_features in [0.2, 0.5, 0.7]:
                    forest_model = RandomForestClassifier(
                        n_estimators=n_trees,
                        bootstrap=True,
                        max_depth=max_depth,
                        max_samples=max_samples,
                        min_samples_leaf=min_leaf,
                        max_features=max_features,
                        random_state=1
                    )
                    scores = cross_val_score(
                        forest_model,
                        X_train,
                        y_train,
                        cv=10,
                        n_jobs=-1
                    )
                    mean_score = scores.mean()
                    results.append({"n_estimators": n_trees,
                                    "max_depth": max_depth,
                                    "min_sample_leaf": min_leaf,
                                    "max_samples": max_samples,
                                    "max_features": max_features,
                                    "accuracy":mean_score
                                    })


best = max(results, key=lambda x: x["accuracy"])

print(best)

best_model = RandomForestClassifier(
                        n_estimators=best["n_estimators"],
                        bootstrap=True,
                        max_depth=best["max_depth"],
                        max_samples=best["max_samples"],
                        min_samples_leaf=best["min_sample_leaf"],
                        max_features=best["max_features"],
                        random_state=1
                    )

best_model.fit(X_train,y_train)
y_forest_preds = best_model.predict(X_test)

print(confusion_matrix(y_test, y_forest_preds))
print()
print(classification_report(y_test, y_forest_preds))
print()
print(best_model.feature_importances_)
```
