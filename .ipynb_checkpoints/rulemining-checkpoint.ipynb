# Business Understanding

The company wants to increase revenue by recommending products
that customers are likely to purchase based on their previous purchases.

# Data Understanding

The dataset contains 100,000 transactions and 20 product groups.
Each ProdX column indicates whether that product group occurred
in the transaction.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_csv('drone_prod_groups.csv', sep=',')
print(df)
```

# Data Preparation

The transaction ID is removed and the product indicators are
converted to Boolean values.

```python
# drop id
df = df.drop(columns='ID')

# replace 1.0 values with True
df = df.apply(lambda x: x.map(lambda y: True if y == 1.0 else False))
df

```

# Modeling

Apriori is used to find frequent itemsets

```python
frequent_itemsets = apriori(df, min_support=0.1, use_colnames=True)
frequent_itemsets
```

```python
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.4)

rules = rules.sort_values(by='confidence', ascending=False)

rules
```

# Evaluation

The rules are evaluated using support, confidence and lift.

The association rules show two strong product relationships: Prod15 → Prod9 and Prod20 → Prod19. Both have high confidence (93.8% and 91.1%) and high lift (4.73 and 4.42), meaning these products are frequently purchased together.

The support for Prod15 and Prod9 is 11.15%, meaning they appear together in 11.15% of all transactions. Prod20 and Prod19 have a higher support of 13.48%, so this combination occurs more frequently.
